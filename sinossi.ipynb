Capstone Project:

Data la mia grande passione per la pesca, desidero sviluppare un applicativo intelligente di supporto ai pescatori.



Il progetto prevede lo sviluppo di un applicativo intelligente in grado di rispondere a quesiti normativi puntuali, contestualizzati rispetto alla posizione geografica esatta dell'utente. Sfruttando la geolocalizzazione GPS e un sistema RAG (Retrieval-Augmented Generation), l'applicazione individua le coordinate del pescatore, recupera automaticamente le ordinanze e i regolamenti pesca locali vigenti per quel determinato bacino, ed elabora risposte chiare, accurate e aggiornate in linguaggio naturale.
